## Find Unique Top-Rated Movies
We'll create a function to find unique movies with an IMDb score above a certain threshold that the user provides. 

```python
def get_unique_top_movies(min_score):
    # Filter for movies with IMDb score above min_score
    high_score_df = df[df['imdb_score'] >= min_score]

    # Extract the 'title' column
    high_score_titles = df.loc[df['imdb_score'] >= min_score, 'title']
    
    # Extract titles
    high_score_titles = set(high_score_titles)

    # Return unique titles
    return list(high_score_titles)
```

```python
# Test the function
print(get_unique_top_movies(9.0)) # Should be the same as the list we got from the previous section.
```

## Find Top Movies from a Specific Decade
We'll create a function to retrieve movies from a particular decade with high IMDb ratings. This function can be useful for generating lits of top-rated movies from different time periods. 

```python
def get_top_movies_from_decade(decade_start, min_score):
    # Filter for movies released within the decade
    decade = df[(df['release_year'] >= decade_start) & (df['release_year'] <= decade_start + 9)]

    # Further filter by IMDb score
    decade = decade[decade['imdb_score'] >= min_score]
    
    # Extract titles
    titles = set(decade['title'])

    # Return unique titles
    return list(titles)
```

```python
# Test the function
print(get_top_movies_from_decade(1990, 8.5))
```

## List All Actors in a Given Title
We'll create a function that takes a title as input and returns the names of all actors in that title, combined into a single string. 

```python
def get_actors_in_title(title):
    title_df = df[(df['title'] == title) & (df['role'] == 'ACTOR')]

    # Extract the 'name' column for actor names
    title_df = title_df['name']

    # Combine actor names into a single string
    actors = ', '.join(title_df.to_list())

    # Return the list of actors
    return actors
```

```python
# Test the function
get_actors_in_title('Taxi Driver')
```

## Categorize Movies by IMDb Score
Let's categorize movies and shows based on their IMDb scores to provide a quick evaluation of their popularity or quality.

**Rating Categories**
- Excellent: IMDb score of 9.0 or higher
- Good: IMDb score between 7.0 and 8.9
- Average: IMDb score between 5.0 and 6.9
- Low: IMDb score below 5.0

```python
def categorize_movies_by_score(title):
    # Filter for the specific title
    title_df = df[df['title'] == title]

    # Check if the title exists in the dataset
    if title_df.empty:
        return "Title not found in the dataset."

    # Build one result per row using that row's name and score
    results = []
    for _, row in title_df.iterrows():
        name = row['name']
        score = row['imdb_score']
        if score >= 9.0:
            results.append(f'{name} - Score: {score} - Excellent')
        elif score >= 7.0 and score < 9.0:
            results.append(f'{name} - Score: {score} - Good')
        elif score >= 5.0 and score < 7.0:
            results.append(f'{name} - Score: {score} - Average')
        else:
            results.append(f'{name} - Score: {score} - Low')

    return results
        
```

```python
# Test the function
print(categorize_movies_by_score('Taxi Driver'))
```
